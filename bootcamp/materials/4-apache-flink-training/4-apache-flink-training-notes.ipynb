What is a streaming pipeline? It's important to define this as there's real time, near real time, etc. A lot of times, stakeholders may say they need data in "real time", but in reality, they may not actually need it in literal real time. As a data engineer, it's important to clarify these things so you don't over engineer solutions. Why do they need it in real time? What are they doing to use the data for? How often are they going to be checking it, etc?

Streaming pipelines process data in a low-latency way. Processes data between minutes to 30 minutes and or intraday. Intraday meaning they're gonna run more that once per day.

- Streaming (or continuous)
    - data is processed as it is generated
    - example: flink
- Near real-time
    - data is processed in small batches every few minutes
    - example: spark structured streaming

Real-time and streaming are often synonymous, but not always. Stakeholders might think in real time but in reality all they need is near real time or even once per day.

Micro-batching == near real time

Real time <> streaming

90% of the time, when stakeholders as for a real time data pipeline, they need near real time, not streaming. Sometimes they actually need it only once per day... so it's really important to clarify the SLA (service level agreement).

Should you even use a streaming pipeline? Don't do it if no one on the team knows about streaming. Also, homogeneity of pipelines is ideal.

Data quality is a lot harder to do in streaming. What if you catch a data quality issue?

Streaming-only use cases:
- detection fraud, preventing bad behavior
- high frequency trading
- live event processing (sports analytics)

These are good use cases because without real time streaming, these analyses would be completely useless. For high frequency trading, if you're data is 2 hours late, you can't do anything with it.

Gray area use-cases:
- data that is served to customers
- reducing the latency of upstream master data

Streaming pipelines are legit a different set of skills so it's important to not take it for granted.

Streaming / batch continuum:

Daily batch --> Hourly batch --> microbatch --> streaming / continuous

hourly / microbatch sometimes called near real time. continuous processing called real time.

Streaming sources:
- kafka
- rabbitmq

Compute engine:
- Flink
- Spark Structured Streaming

The destination:
- "the sink"
    - another kafka topic
    - iceberg
    - postgres

Streaming challenges:
- out of order events (newer data ahead of older data, data is not in the right order)
- late arriving data
- recovering from failures
