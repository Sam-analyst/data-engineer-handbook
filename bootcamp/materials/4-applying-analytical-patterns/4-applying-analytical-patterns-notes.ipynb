# Applying Analytical Patterns: Exploring SQL, Scaling Projects and Aggregation Analysis Day 1 Lecture

In this lecture, he's going to walk through 2 common analytical patterns, growth accounting (like daily active users at facebook) and survivorship analysis, like what percent of people that sign up are still active. With data engineering, there are only a handful of patterns for which you'll base you model off of.

Understanding what type of analysis (repeatable pattern) is crucial. If you can identify what you're trying to do at a high level, it becomes far more easier on what you need to code.

There are 3 common patterns:
- aggregation based patterns
- cumulation based patterns
- window based patterns

Aggregation patterns:
- Trend analysis
- Root cause analysis

Pretty much gonna be doing group by, count, avg, etc.

Most common pattern. Probs like 50%.

He also talked about how its not ideal to aggregate at the fact level. He said its ok to have a separate table that's aggregated up to some level (daily) to reduce row count.

Cumulation-based patterns
- state transition tracking
- retention (also called j curves)

Full outer join is your best friend here when it comes to building pipelines. 

Growth accounting (new, retained, churned, resurrected, stale). He really emphasized this pattern. He used it at facebook, netflix and airbnb. Super helpful to track and monitor the health of machine learning models.

Survivor Analysis J-curves (almost like an L curve). But this shows overtime, retention falls. Y axis is retention, X axis is time.

Window-based Analyses

- Rolling Sum / Average
- Ranking

Rolling is the key word here.
